## ***4. EDA Process***

#### 1. Does gender matter for extending policies?

````python
gender_mean = df.groupby('sex')['charges in INR'].mean()
print("Average Charges by Gender:\n", gender_mean, "\n")

````

#####  Q1. Does gender matter for extending policies?

Insight Found:
Gender does not show a major difference in claim amounts. Both male and female policyholders claim almost similar insurance amounts.

Positive Impact:
This insight helps confirm that gender-based policy pricing is not required, ensuring fairness and reducing unnecessary discrimination.

#### 2. What is the average amount spent per policy?

````python
avg_spent = df['charges in INR'].mean()
print("Average charges per policy:", avg_spent, "\n")
````

##### Q2. What is the average amount spent per policy?

Insight Found:
The average claim amount is approximately ₹13,000 per policy, giving the company a baseline spending metric.

Positive Impact:
This helps the insurer understand expected average expenditure per customer and assists in policy budgeting and pricing decisions.

#### 3. Do claims vary by geographic location?

````python
region_mean = df.groupby('region')['charges in INR'].mean()
print("Average charges by region:\n", region_mean, "\n")
````

#####  Q3. Should the company offer separate policies based on region?

Insight Found:
Claim amounts are evenly distributed across all four regional categories. No region shows unusually high or low claim values.

Positive Impact:
This avoids unnecessary geographic segmentation and simplifies policy design while keeping processing uniform across regions.

#### 4. Does number of dependents make a difference in claim amounts?

````python
children_mean = df.groupby('children')['charges in INR'].mean()
print("Average charges by number of children:\n", children_mean, "\n")
````

##### Q4. Does number of dependents make a difference in claim amounts?

Insight Found:
The number of dependents (children) does not significantly change the claim amounts. Claim behavior remains similar across different dependent counts.

Positive Impact:
This prevents the company from overpricing or altering policies based on dependents, making policies simpler and more customer-friendly.

#### 5. Does BMI give insights for claims?

````python
bmi_correlation = df['bmi'].corr(df['charges in INR'])
print("Correlation between BMI and charges:", bmi_correlation, "\n")
````

##### Q5. Does BMI provide any idea for claim levels?

Insight Found:
Higher BMI levels correlate with increased claim amounts. Obesity and overweight categories tend to have higher medical risk.

Positive Impact:
This enables the insurer to consider BMI as a health-risk indicator and design health-based incentives or preventive programs.

#### 6. Is smoker status important for claim?

````python
smoker_mean = df.groupby('smoker')['charges in INR'].mean()
print("Average charges by smoker status:\n", smoker_mean, "\n")
````

##### Q6. Is smoker status important in insurance claims?

Insight Found:
Smokers have significantly higher average claim amounts compared to non-smokers. Smoking is a strong risk factor.

Positive Impact:
This allows the company to evaluate premiums more accurately and promotes a shift towards healthier lifestyles through incentives.

#### 7. Does age affect claim?

````python
age_correlation = df['age'].corr(df['charges in INR'])
print("Correlation between age and charges:", age_correlation, "\n")
````

##### Q7. Does age affect claim amount?

Insight Found:
Age shows a positive correlation with claim values. Older policyholders tend to claim more due to higher medical risk.

Positive Impact:
Age-based risk understanding supports accurate underwriting and better policy pricing aligned with customer risk levels.

#### 8. Can discounts be extended based on BMI?

````python
bmi_mean = df['bmi'].mean()
print("Average BMI:", bmi_mean)

# Segmenting based on BMI ranges (example logic)
df['BMI_Category'] = pd.cut(df['bmi'],
                            bins=[0, 18.5, 24.9, 29.9, 100],
                            labels=['Underweight', 'Normal', 'Overweight', 'Obese'])

bmi_category_mean = df.groupby('BMI_Category')['charges in INR'].mean()
print("\nAverage charges by BMI category:\n", bmi_category_mean)
````

##### Q8. Can the company extend discounts based on BMI?

Insight Found:
Normal BMI customers have lower claim amounts than overweight/obese categories. Healthy BMI leads to lower expenditure.

Positive Impact:
This enables wellness-based discount models and encourages healthy behavior, reducing long-term medical costs for both insurer and customers.
